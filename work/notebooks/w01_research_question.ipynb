Search question: Which pages cover the same topic even when worded differently, and which topics have demand but no page? Unit of analysis: a single page on a client's site. Output: a topic cluster per page + list of overlap topics and gap topics. Decision it improves: whether to merge competing pages or create new content for underserved topics. Action someone takes: SEO/content team merges overlapping pages, or greenlights new content for a gap topic. Cost of a wrong call: false overlap → merging/deleting content that served a real distinct search intent, losing traffic; false gap → wasted effort building content nobody searches for. Why ML helps: manually judging topic similarity across hundreds of pages doesn't scale; embeddings + clustering surface patterns for a human to validate — not full automation.
