My lane (Lane 3) is a **clustering** task — unsupervised learning. There's no single
label to predict; the goal is to group content pages into naturally occurring
performance archetypes based on their metrics, then interpret what each cluster

Since this is unsupervised, "accuracy" doesn't apply. Two checks:
1. Silhouette score — a number I can defend, roughly 0 = overlapping clusters,
   1 = clean separation. Realistic target on noisy real-world data: >0.2-0.3.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
```

A fixed rule (e.g. "if ctr < X and position > Y then archetype A") only handles 1-2
dimensions at a time and needs someone to hand-pick every threshold. Content
performance is genuinely multivariate — a page can have high impressions but low
engagement, or good position but stale content, in combinations that don't reduce
to a simple checklist. Across 30,000 pages and 32 clients, no one can eyeball every
combination. Clustering lets the natural groupings emerge from the data across many
dimensions at once — exactly what if/else rules can't do without turning into an
unmanageable nested mess of thresholds.
